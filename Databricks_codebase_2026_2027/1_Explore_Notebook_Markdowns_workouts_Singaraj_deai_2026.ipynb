###### How to bold content

1. `**bold content**` — wrap the text in double stars, before and after
2. `<b>bold content</b>` — use HTML tags
3. `__bold content__` — double underscores also bold the text (another method)

###### Italics

*Italics* — prefix and suffix the text with a single star (`*`).

###### How to italicize content

- *Stars* — use `*` before and after the word
- _Underscores_ — use `_` before and after the word (also renders as italics)

###### How to embed URLs in a notebook

###### Markdown Tables

A table in markdown is created using **pipes** `|` and **hyphens** `-`:

- Each **row** is written on a new line, with cells separated by `|` (pipe)
- The **first row** defines the column headers (here: `col1`, `col2`)
- The **second row** (`|------|------|`) is the separator — it draws the line under the header and marks where the header ends and the body begins
- Every following line is a **data row** (here: `a | b` and `c | d`)

**Example table:**

| col1 | col2 |
|------|------|
| a | b |
| c | d |

**Rendered result:**

- A table with **2 columns** (`col1` and `col2`) and **2 data rows**
- Row 1: `a` under `col1`, `b` under `col2`
- Row 2: `c` under `col1`, `d` under `col2`
